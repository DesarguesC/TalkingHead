```python
try:
    
    if scene:
        print(f"模型包含 {len(scene.meshes)} 个网格。")
        for mesh in scene.meshes:
            print(f"  网格名称: {mesh.name}")
            print(f"    顶点数量: {len(mesh.vertices)}")
            print(f"    面数量: {len(mesh.faces)}")
            
    else:
        print("加载 FBX 文件失败。")
except pyassimp.AssimpError as e:
    print(f"Assimp 错误: {e}")
finally:
    pyassimp.release(scene)
```
